```python
import pandas as pd
data = {'Name':['Alice', 'Divya'], 'age':[24,25], 'Salary': [50000, 60000]}
df = pd.DataFrame(data)
print("df:\n", df)
#first few rows of the data frame
print(df.head((5)))
print(df.head(1))


#display last few rows
print("tail :\n", df.tail(1))
print("tail :\n", df.tail(4))



```

```python
# Purpose: Provides a concise summary of the DataFrame.
print(df.info())

# Purpose: Generates descriptive statistics of the DataFrame’s numerical columns.
print(df.describe())
```

```python
# filtering data
print(df[df['age']> 20], end="\n\n")
print(df[df['age'].between(23, 24)], end="\n\n")
print(df[df['Name'].str.contains("iv")], end="\n\n")
print(df[df['Name'].str.startswith("Al")], end="\n\n")
print(df[df['Name'].str.endswith("ya")], end="\n\n")

```

```python
df.query('age > 24')  # Same as df[df['age'] > 24]
```

```python
df[df['Name'].isin(['Alice', 'Bob'])]  # Selects rows where 'Name' is 'Alice' or 'Bob'
```

```python
# df['New_Salary'] = df['Salary']*1.1
new_salary = df['Salary']*1.1

df.insert(1, 'New_Salary', new_salary)
print(df)
```

```python
df.drop('New_Salary', axis=1, inplace=True)
print(df)
```

```python
df.rename(columns={"Salary":"New_Salary"}, inplace=True)
print(df)
```

```python
df.rename(index={0: "Row_0"}, inplace=True)
print(df)
```

```python
df.drop(1, axis=0, inplace=True)
print(df)
```
