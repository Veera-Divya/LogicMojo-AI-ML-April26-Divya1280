```python
# Question: Stack arrays a and b vertically
# input: a = np.arange(10).reshape(2,-1)
#        b = np.repeat(1, 10).reshape(2,-1)
import numpy as np
a = np.arange(10).reshape(2,-1)
b = np.repeat(1,10).reshape(2, -1)
print("a :", a)
print("b :", b)
print(np.vstack([a,b]))
print(np.concatenate([a,b], axis = 0))
```

```python
### reverse an array
arr = np.array([1,2,3,4,5,6])
print(arr[::-1])
```

```python
# Question: Get all items between 5 and 10 from a.

# Input: a = np.array([2, 6, 1, 9, 10, 3, 27])

a = np.array([2, 6, 1, 9, 10, 3, 27])
print(a[(a>5) & (a<10)])
```

```python
# Find the maximum value from both of these array

a = np.array([5, 7, 9, 8, 6, 4, 5])
b = np.array([6, 3, 4, 8, 9, 7, 1])
print(max(max(a), max(b)))
```
