Array Reshaping -> Reshape, Ravel, flatten

```python
arr = np.arange(1,10)
new_arr= arr.reshape(3,3)
new_arr
```

```python
# ravel -> convert the array to 1 D array
# return a view  of original array which basically means any changes you apply here reflected in original array
new_arr2= np.ravel(new_arr)
print(new_arr2)
new_arr2[0] = 1
print(new_arr)
```

```python
# flaten -convert the array to 1 D array
# return the copy of array
flat = new_arr.flatten()
print(flat)
flat[0] = 0
print(new_arr)
```

Index Arrays


```python
arr = np.array([1,2,3,4,5,6])
ind = [0,2]
print(np.take(arr, ind))
```

```python
# Iterating  with nditer()
arr = np.array([[1,2,3], [4,5,6]])
for x in np.nditer(arr):
  print(x, end=" ")
```

```python
# Iterating  with ndenumerate
arr = np.array([[1,2,3], [4,5,6]])
for i, x in np.ndenumerate(arr):
  print(i, x)
```

```python
# Views vs Copies

#view
arr = np.array([1,2,4,5,6])
view = arr[1:3]
view[0] = 200
print(arr)
```

```python
#copy
arr = np.array([1,2,4,5,6])
copy = arr[1:3].copy()
copy[0] = 200
print(arr)
```

```python
# Transpose
arr = np.array([[1,2,3], [4,5,6]])
print(arr)
print(arr.T)
print(arr)
```

```python
# swapaxes -> swap 2 specific axes in a matrix
swap = np.swapaxes(arr, 0, 1)
print(swap)
swap.shape
```
